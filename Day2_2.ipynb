```python
census = pd.read_csv("census.csv", index_col = 0)
census
```

```python
census.groupby("STNAME")["CTYNAME"].count().idxmax()
```

```python
census.groupby("STNAME") \
      .apply(lambda x: x.nlargest(3, "CENSUS2010POP")["CENSUS2010POP"].sum()) \
      .nlargest(3)
```

```python
x = census.groupby('STNAME').size()
ans = x.idxmax()
print(ans)
```

```python
top_contries_df = census.df.sort_values(by=['STNAME', 'CENSUS2010POP'], ascending=False).groupby('STNAME').head(3)
ans = top_contries_df.groupby('STNAME').sum().sort_values(by='CENSUS2010POP', ascending=False).head(3).index.tolist()
print(ans)
```

```python
top_counties_df = census.sort_values(by=['STNAME',"CENSUS2010POP"],
                                         ascending=False).groupby('STNAME').head(3)
ans = top_counties_df.groupby('STNAME').sum().sort_values(by='CENSUS2010POP',
                                                          ascending=False).head(3).index.tolist()
print(ans)
```
